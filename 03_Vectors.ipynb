# Q6 - 10 mins) Depreciation Schedule

A company buys a machine for **120,000 TL**. After **5 years** it can be sold for **20,000 TL** (salvage value). Two depreciation methods are given:

$$\text{Straight-line (SL)}_t = \frac{\text{Cost} - \text{Salvage}}{\text{Life}} \qquad\qquad \text{Sum-of-years'-digits (SYD)}_t = (\text{Cost} - \text{Salvage}) \times \frac{\text{Life} - t + 1}{1 + 2 + \dots + \text{Life}}$$

$$\text{Book value at end of year } t = \text{Cost} - \text{accumulated depreciation up to year } t$$

a) Create `cost`, `salvage`, `life` and `years <- 1:life`.

b) Create `dep_sl`: the same SL amount repeated for every year (use `rep()`).

c) Create `dep_syd` in one line using the descending sequence `life:1` (for $t = 1,\dots,5$, the factor $\text{Life}-t+1$ is $5,4,3,2,1$) and `sum(1:life)`.

d) Compute the book values `book_sl` and `book_syd`. (Hint: `cumsum(c(1, 2, 3))` gives `1 3 6`, a running total.)

e) In which years is SYD depreciation **lower** than SL? (logical subsetting on `years`)

f) Check that the last SYD book value equals the salvage value (use `round(x, 2)`).

```r
# ANSWER
cost <- 120000
salvage <- 20000
life <- 5
years <- 1:life

dep_sl  <- rep((cost - salvage) / life, times = life)          # 20000 each year
dep_syd <- (cost - salvage) * (life:1) / sum(1:life)           # 33333.3 26666.7 20000 13333.3 6666.7

book_sl  <- cost - cumsum(dep_sl)                              # 100000 80000 60000 40000 20000
book_syd <- cost - cumsum(dep_syd)                             # 86666.7 60000 40000 26666.7 20000

years[dep_syd < dep_sl]                                        # 4 5

round(book_syd[life], 2) == salvage                            # TRUE
```

# Q7 - 10 mins) Break-Even (CVP) Analysis

A bakery sells a product at **250 TL** per unit. The variable cost is **150 TL** per unit and the monthly fixed cost is **80,000 TL**.

$$\text{Profit} = \text{Units} \times \text{Price} - (\text{Fixed Cost} + \text{Units} \times \text{Variable Cost})$$

$$\text{Break-even units} = \frac{\text{Fixed Cost}}{\text{Price} - \text{Variable Cost}} \qquad \text{Margin of safety (\%)} = \frac{\text{Actual Units} - \text{Break-even Units}}{\text{Actual Units}} \times 100$$

a) Create `units` from 0 to 1600 in steps of 200 with `seq()`.

b) Compute the vectors `revenue`, `total_cost` and `profit` for all volumes at once.

c) At which sales volumes does the bakery make a **loss**? (logical subsetting on `units`)

d) Compute `bep` with the formula, then check that profit at that volume is zero: `profit[units == bep]`.

e) If actual sales are 1,200 units, what is the margin of safety?

f) How many units must be sold to earn a **target profit of 50,000 TL**? (Hint: add the target to the fixed cost in the break-even formula.)

```r
# ANSWER
price <- 250
vc <- 150
fc <- 80000

units <- seq(0, 1600, by = 200)

revenue    <- units * price
total_cost <- fc + units * vc
profit     <- revenue - total_cost         # -80000 -60000 -40000 -20000 0 20000 40000 60000 80000

units[profit < 0]                          # 0 200 400 600

bep <- fc / (price - vc)                   # 800
profit[units == bep]                       # 0

(1200 - bep) / 1200 * 100                  # 33.33

(fc + 50000) / (price - vc)                # 1300
```

# Q8 - 10 mins)  Budget vs. Actual Report

Six departments have the following budgets and actual spending (in TL). **R&D has not sent its report yet**, so its actual value is missing (`NA`).

|---|---|---|---|---|---|---|
| Budget | 50000 | 120000 | 30000 | 45000 | 70000 | 40000 |
| Actual | 56000 | 114000 | 31500 | NA | 77000 | 38000 |

$$\text{Variance} = \text{Actual} - \text{Budget} \quad(\text{positive} = \text{overspending}) \qquad\qquad \text{Variance \%} = \frac{\text{Variance}}{\text{Budget}} \times 100$$

a) Create the vectors `dept`, `budget`, `actual` and compute `variance` and `variance_pct` (rounded to 1 decimal).

b) Run `dept[variance > 0]`. What do you notice about `NA`? Then fix it using `which()`: `dept[which(variance > 0)]` gives the overspending departments.

```r
# ANSWER
dept   <- c("Marketing", "Production", "HR", "R&D", "Sales", "IT")
budget <- c(50000, 120000, 30000, 45000, 70000, 40000)
actual <- c(56000, 114000, 31500, NA, 77000, 38000)

variance     <- actual - budget
variance_pct <- round(variance / budget * 100, 1)     # 12 -5 5 NA 10 -5

dept[variance > 0]                # contains an NA entry: a logical NA index returns NA
dept[which(variance > 0)]         # "Marketing" "HR" "Sales"

sort(variance, decreasing = TRUE, na.last = TRUE)     # 7000 6000 1500 -2000 -6000 NA

dept[order(variance, decreasing = TRUE, na.last = NA)]  # "Sales" "Marketing" "HR" "IT" "Production"

actual[4] <- 47000
variance <- actual - budget
sum(variance)                     # 8500
```

# Q10 - 10 mins) Market Equilibrium & Price Ceiling

A market has the following demand and supply curves ($P$ = price in TL):

$$Q_d = 100 - 2P \qquad\qquad Q_s = -20 + 4P$$

The market is in equilibrium where $Q_d = Q_s$. If $Q_d > Q_s$ there is a **shortage**, if $Q_d < Q_s$ there is a **surplus**.

a) Create `price` from 10 to 30 in steps of 2.5 with `seq()`.

b) Compute `qd` and `qs` for all prices at once.

c) At which prices is there a **shortage**?

d) Find the equilibrium price and quantity. (Hint: the gap `qd - qs` is closest to zero there, so use `which.min(abs(qd - qs))`.)

e) The government sets a **price ceiling of 15 TL**. How big is the shortage at 15 TL? (Hint: use `price == 15` to pick the right element.)

f) Seller revenue is $P \times Q$. At 15 TL sellers only sell $Q_s$. Compare the revenue at the ceiling with the revenue at equilibrium.

```r
# ANSWER
price <- seq(10, 30, by = 2.5)

qd <- 100 - 2 * price
qs <- -20 + 4 * price

price[qd > qs]                          # 10 12.5 15 17.5

idx <- which.min(abs(qd - qs))
eq_price <- price[idx]                  # 20
eq_qty   <- qd[idx]                     # 60

(qd - qs)[price == 15]                  # shortage of 30 units

rev_ceiling <- 15 * qs[price == 15]     # 600
rev_eq      <- eq_price * eq_qty        # 1200
c(ceiling = rev_ceiling, equilibrium = rev_eq)
```

# Q11 - 15 mins)  Growth and Inflation

The nominal GDP (billion TL) and the GDP deflator (base year 2019 = 100) of a country are given below.

| Year | 2019 | 2020 | 2021 | 2022 | 2023 | 2024 |
|---|---|---|---|---|---|---|
| Nominal GDP | 1000 | 1080 | 1170 | 1250 | 1360 | 1480 |
| GDP deflator | 100 | 103 | 107 | 110 | 115 | 119 |

$$\text{Real GDP} = \frac{\text{Nominal GDP}}{\text{Deflator}} \times 100 \qquad \text{Growth}_t = \frac{\text{Real GDP}_t - \text{Real GDP}_{t-1}}{\text{Real GDP}_{t-1}} \times 100 \qquad \text{Inflation}_t = \frac{\text{Deflator}_t - \text{Deflator}_{t-1}}{\text{Deflator}_{t-1}} \times 100$$

a) Create `years` (use `2019:2024`), `nominal` and `deflator`, then compute `real`.

b) Compute `growth` using `diff()` for the numerator and negative indexing `real[-n]` for the denominator (`n <- length(real)`, the previous year's value). Round to 2 decimals.

c) Compute `inflation` the same way from the deflator.

d) Growth and inflation start from the **second** year, so create `growth_years <- years[-1]`. In which year was real growth the highest?

e) In which years did real growth **exceed** inflation?

f) What is the average real growth rate?

```r
# ANSWER
years    <- 2019:2024
nominal  <- c(1000, 1080, 1170, 1250, 1360, 1480)
deflator <- c(100, 103, 107, 110, 115, 119)

real <- nominal / deflator * 100           # 1000 1048.54 1093.46 1136.36 1182.61 1243.70

n <- length(real)
growth    <- round(diff(real)     / real[-n]     * 100, 2)   # 4.85 4.28 3.92 4.07 5.17
inflation <- round(diff(deflator) / deflator[-n] * 100, 2)   # 3.00 3.88 2.80 4.55 3.48

growth_years <- years[-1]
growth_years[which.max(growth)]            # 2024

growth_years[growth > inflation]           # 2020 2021 2022 2024

mean(growth)                               # 4.458
```

# Q12 - 15 mins) BONUS -  Is This Project Worth It?

A project needs an investment of **100,000 TL** today (year 0) and returns **30,000, 40,000, 50,000, 40,000 TL** at the end of years 1-4. The discount rate is **10%**.

$$PV_t = \frac{CF_t}{(1 + r)^t} \qquad\qquad NPV = \sum_{t=0}^{4} PV_t$$

The project is accepted if $NPV > 0$.

a) Create `cf` (the investment is **negative**), `t <- 0:4` and `r <- 0.10`.

b) Compute the vector `pv` of present values, then `npv <- sum(pv)`.

c) Compute the running total `cumsum(pv)`. In which year does it first become positive (discounted payback)? (Hint: `t[which(cumsum(pv) >= 0)[1]]`)

d) Change `r` to **0.20** and then **0.25** and rerun your code. What happens to the NPV, would you still accept the project?

```r
# ANSWER
cf <- c(-100000, 30000, 40000, 50000, 40000)
t  <- 0:4
r  <- 0.10

pv  <- cf / (1 + r)^t
npv <- sum(pv)                             # 25216.86
npv

cumsum(pv)                                 # -100000 -72727 -39669 -2104 25217
t[which(cumsum(pv) >= 0)[1]]               # 4

# r = 0.20 -> NPV = 1003.09 (still accepted, barely)
# r = 0.25 -> NPV = -8416    (rejected)
```
