**EXERCISE 6:**
Create a sample of uniformly generated numbers between 0 and 1 with size 8! = 40320. Assign to `run1`.
Now create vectors `run2` to `run8` where `runN` is a vector of 8! / N items, each a sum of N uniformly generated numbers.

1) Plot the histograms of run1 to run8, do you see the pattern?
2) Get the five-point statistical summaries of run1 to run8 using `summary()`.

```python
%%R
# SOLUTION 6 (SHORT WAY):
runcount <- factorial(8)
run1 <- runif(runcount)
run2 <- replicate(runcount / 2, sum(runif(2)))
run3 <- replicate(runcount / 3, sum(runif(3)))
run4 <- replicate(runcount / 4, sum(runif(4)))
run5 <- replicate(runcount / 5, sum(runif(5)))
run6 <- replicate(runcount / 6, sum(runif(6)))
run7 <- replicate(runcount / 7, sum(runif(7)))
run8 <- replicate(runcount / 8, sum(runif(8)))

hist(run1)
hist(run8)
summary(run1)
summary(run8)
```

**EXERCISE 7:**
Randomly assign numbers 1 to 5 to a sample of 1000 cases (categories).
Create normally distributed numbers with standard deviation equal to one-third of the category value, and mean equal to half of the category value.

```python
%%R
# SOLUTION 7:
categories <- sample(5, 10000, replace = TRUE)
zscores <- rnorm(10000)
values <- categories/2 + zscores * categories/3

hist(values)
hist(values[categories == 1])
hist(values[categories == 5])
```

**EXERCISE 8:**
Create a sample of 5000 logarithmic returns with a mean of 0.05 and sd of 0.5. Plot its histogram.
Calculate the price series starting with 1, find the geometric returns, and plot their histogram.

```python
%%R
# SOLUTION 8:
returns_ln <- rnorm(5000, mean = 0.05, sd = 0.5)
hist(returns_ln)

prices <- c(1, cumprod(exp(returns_ln)))
returns_geo <- prices[-1] / prices[-length(prices)] - 1
hist(returns_geo)

summary(returns_ln)
summary(returns_geo)
```
