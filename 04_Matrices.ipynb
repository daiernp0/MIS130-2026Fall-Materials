# Q3  — The General Ledger as a Debit/Credit Matrix

In **double-entry bookkeeping**, every transaction is written to (at least) two accounts: one is **debited**, one is **credited**, with the **same amount**. You do *not* need to know the rules — here they are:

| Account type | Accounts in our ledger | Increases with | Decreases with |
|---|---|---|---|
| Asset | Cash, Inventory, Equipment | Debit | Credit |
| Liability | AccountsPayable, BankLoan | Credit | Debit |
| Equity | Capital | Credit | Debit |
| Revenue | Revenue | Credit | Debit |
| Expense | RentExpense | Debit | Credit |

**Boğaziçi Coffee Co.** had these transactions in its first month (journal entries are given):

| # | Event | Debit | Credit | Amount (TL) |
|---|---|---|---|---|
| 1 | Owner invests cash | Cash | Capital | 100,000 |
| 2 | Bank loan received | Cash | BankLoan | 50,000 |
| 3 | Espresso machine bought with cash | Equipment | Cash | 60,000 |
| 4 | Coffee beans bought on credit | Inventory | AccountsPayable | 20,000 |
| 5 | Coffee sold for cash | Cash | Revenue | 35,000 |
| 6 | Rent paid | RentExpense | Cash | 8,000 |
| 7 | Part of supplier debt paid | AccountsPayable | Cash | 12,000 |

a) Create an **8 × 2** matrix `ledger` full of zeros. Rows = the 8 accounts above (in table order), columns = `"Debit"`, `"Credit"`.

b) Post all 7 transactions by **adding** to the right cells, e.g.
`ledger["Cash", "Debit"] <- ledger["Cash", "Debit"] + 100000`

c) **Trial balance:** total debits must equal total credits. Check with `sum()` and `==`.

d) Compute each account's balance: `balance = Debit − Credit`. (Positive = debit balance, negative = credit balance.) How much cash does the company have?

e) Which accounts have a **credit** balance? (Use a logical condition inside `[ ]`.)

f) **Accounting equation check.** Using the balances:

$$\text{Assets} = \text{Liabilities} + \text{Capital} + (\text{Revenue} - \text{Expenses})$$

Remember that liabilities, capital and revenue come out **negative** in `balance`, so flip their sign with `-`. Does the equation hold?



````r
# Your code here
````

````r
# ===================== ANSWER =====================
# a)
ledger <- matrix(0, nrow = 8, ncol = 2)
rownames(ledger) <- c("Cash", "Inventory", "Equipment", "AccountsPayable",
                      "BankLoan", "Capital", "Revenue", "RentExpense")
colnames(ledger) <- c("Debit", "Credit")

# b)
# 1) Owner invests cash
ledger["Cash", "Debit"]               <- ledger["Cash", "Debit"]               + 100000
ledger["Capital", "Credit"]           <- ledger["Capital", "Credit"]           + 100000
# 2) Bank loan
ledger["Cash", "Debit"]               <- ledger["Cash", "Debit"]               + 50000
ledger["BankLoan", "Credit"]          <- ledger["BankLoan", "Credit"]          + 50000
# 3) Equipment for cash
ledger["Equipment", "Debit"]          <- ledger["Equipment", "Debit"]          + 60000
ledger["Cash", "Credit"]              <- ledger["Cash", "Credit"]              + 60000
# 4) Inventory on credit
ledger["Inventory", "Debit"]          <- ledger["Inventory", "Debit"]          + 20000
ledger["AccountsPayable", "Credit"]   <- ledger["AccountsPayable", "Credit"]   + 20000
# 5) Cash sales
ledger["Cash", "Debit"]               <- ledger["Cash", "Debit"]               + 35000
ledger["Revenue", "Credit"]           <- ledger["Revenue", "Credit"]           + 35000
# 6) Rent
ledger["RentExpense", "Debit"]        <- ledger["RentExpense", "Debit"]        + 8000
ledger["Cash", "Credit"]              <- ledger["Cash", "Credit"]              + 8000
# 7) Pay supplier
ledger["AccountsPayable", "Debit"]    <- ledger["AccountsPayable", "Debit"]    + 12000
ledger["Cash", "Credit"]              <- ledger["Cash", "Credit"]              + 12000
ledger

# c)
sum(ledger[, "Debit"])
sum(ledger[, "Credit"])
sum(ledger[, "Debit"]) == sum(ledger[, "Credit"])    # TRUE (285,000 each)

# d)
balance <- ledger[, "Debit"] - ledger[, "Credit"]
balance
balance["Cash"]                                      # 105,000

# e)
names(balance)[balance < 0]

# f)
assets      <- sum(balance[c("Cash", "Inventory", "Equipment")])
liabilities <- -sum(balance[c("AccountsPayable", "BankLoan")])
capital     <- -balance["Capital"]
net_income  <- -balance["Revenue"] - balance["RentExpense"]
assets
liabilities + capital + net_income
assets == liabilities + capital + net_income         # TRUE (185,000)
````

# Q4  — Auditing: Catching an Intern's Bookkeeping Errors

An intern typed 8 journal entries into a matrix. Each row is one transaction; column 1 = amount debited, column 2 = amount credited. In a correct entry the two are **equal**.

| | Debit | Credit |
|---|---|---|
| T1 | 100000 | 100000 |
| T2 | 50000 | 50000 |
| T3 | 60000 | 6000 |
| T4 | 20000 | 20000 |
| T5 | 35000 | 53000 |
| T6 | 8000 | 8000 |
| T7 | 12000 | 12000 |
| T8 | 4500 | 4500 |

a) Create `journal` (use `byrow = TRUE`), with row names `T1 … T8` made with `paste()` and column names `Debit`, `Credit`.

b) Create a logical vector `balanced` that is `TRUE` when debit equals credit. Use `all()` to ask "is everything fine?" and `which(!balanced)` to find the broken entries.

c) Compute the difference (Debit − Credit) **only for the broken rows**.

d) **Auditor's rule of thumb:** if the difference is divisible by **9**, the error is probably a *transposition* (35000 ↔ 53000) or a *slide* (a missing zero). Test this with `%%`.

e) **Audit threshold:** every amount above 50,000 TL must be reviewed by a manager. Create the logical matrix `journal > 50000`, count how many cells exceed it with `sum()`, and find their positions with `which(..., arr.ind = TRUE)`.

f) Fix the two errors (the correct amounts are 60000 and 35000) by overwriting the wrong cells, then confirm `all(...)` is `TRUE`.

````r
# Your code here
````

````r
# ===================== ANSWER =====================
# a)
journal <- matrix(c(100000, 100000,
                     50000,  50000,
                     60000,   6000,
                     20000,  20000,
                     35000,  53000,
                      8000,   8000,
                     12000,  12000,
                      4500,   4500),
                  ncol = 2, byrow = TRUE)
rownames(journal) <- paste("T", 1:8, sep = "")
colnames(journal) <- c("Debit", "Credit")
journal

# b)
balanced <- journal[, "Debit"] == journal[, "Credit"]
balanced
all(balanced)           # FALSE
which(!balanced)        # T3, T5

# c)
diffs <- journal[!balanced, "Debit"] - journal[!balanced, "Credit"]
diffs                   # 54000, -18000

# d)
diffs %% 9 == 0         # TRUE TRUE -> classic slide / transposition errors

# e)
big <- journal > 50000
big
sum(big)                       # 4 cells
which(big, arr.ind = TRUE)

# f)
journal["T3", "Credit"] <- 60000
journal["T5", "Credit"] <- 35000
all(journal[, "Debit"] == journal[, "Credit"])   # TRUE
````

- Operating profit = Total CM (all products) − Fixed costs

f) **What-if:** butter prices jump and Börek's material cost rises by **25%**. Overwrite that single cell and recompute the operating profit. How much profit is lost?

````r
# Your code here
````

````r
# ===================== ANSWER =====================
# a)
cost <- cbind(Materials   = c(4, 6, 15),
              Labor       = c(3, 4, 8),
              VarOverhead = c(1, 2, 3))
rownames(cost) <- c("Simit", "Pogaca", "Borek")
cost

units <- c(6000, 3000, 1200)
price <- c(15, 20, 45)
fixed <- 60000

# b)
unit_vc <- rowSums(cost)          # 8, 12, 26
cm      <- price - unit_vc        # 7, 8, 19
cm
names(cm)[which.max(cm)]          # "Borek"

# c)
total_cost <- cost * units
total_cost
colSums(total_cost)["Materials"]  # 60,000

# d)
total_cm <- sum(cm * units)       # 88,800
profit   <- total_cm - fixed      # 28,800
profit

# e)
ceiling(fixed / cm["Simit"])      # 8572 simit

# f)
cost_new <- cost
cost_new["Borek", "Materials"] <- cost_new["Borek", "Materials"] * 1.25
cm_new     <- price - rowSums(cost_new)
profit_new <- sum(cm_new * units) - fixed
profit_new                        # 24,300
profit - profit_new               # 4,500 lost
````

# Q6  —  A Pricing Game Between Two Cafés

Two cafés on campus, **A** and **B**, each choose a `High` or `Low` price. Their monthly profits (thousand TL) depend on **both** choices:

**Café A's profit** (rows = A's choice, columns = B's choice)

  - A's best response to B's choice → look **down a column** of `payA`.
  - B's best response to A's choice → look **along a row** of `payB`.
- **Nash equilibrium:** a cell where **both** players are playing a best response at the same time.

b) If B plays `High`, what is A's best response? What if B plays `Low`? (Hint: `rownames(payA)[which.max(payA[, "High"])]`)

- `bestB`: `TRUE` where `payB` equals the **maximum of its row** (think about how to build the row-max matrix — `byrow` or not?)

Then `nash <- bestA & bestB` and locate it with `which(nash, arr.ind = TRUE)`.

e) Compute `total <- payA + payB`. Which cell gives the cafés the highest **joint** profit? Is it the Nash equilibrium? (This is the famous *Prisoner's Dilemma*.)

> **Class activity:** Pair up. Each person secretly writes `High` or `Low` on paper, then reveal. The TA records the class results in a 2 × 2 count matrix. Did the class end up at the Nash equilibrium?

````r
# Your code here
````

````r
# ===================== ANSWER =====================
# a)
payA <- matrix(c(50, 10,
                 70, 30), nrow = 2, byrow = TRUE)
payB <- matrix(c(50, 70,
                 10, 30), nrow = 2, byrow = TRUE)
rownames(payA) <- rownames(payB) <- c("High", "Low")   # A's choice
colnames(payA) <- colnames(payB) <- c("High", "Low")   # B's choice
payA
payB

# b)
rownames(payA)[which.max(payA[, "High"])]   # "Low"
rownames(payA)[which.max(payA[, "Low"])]    # "Low"

# c)
colnames(payB)[which.max(payB["High", ])]   # "Low"

# d)
colmaxA <- matrix(c(max(payA[, 1]), max(payA[, 2])), nrow = 2, ncol = 2, byrow = TRUE)
bestA   <- payA == colmaxA

rowmaxB <- matrix(c(max(payB[1, ]), max(payB[2, ])), nrow = 2, ncol = 2)  # filled by column -> each row gets its own max
bestB   <- payB == rowmaxB

bestA
bestB
nash <- bestA & bestB
nash
which(nash, arr.ind = TRUE)                 # (Low, Low)

# e)
total <- payA + payB
total
which(total == max(total), arr.ind = TRUE)  # (High, High) = 100 -> better for both, but not a Nash equilibrium
````

$$\sigma = \sqrt{\text{variance}}$$

Use `diag()` and `sqrt()` to get the volatility of each stock, check one with `sd()`, and name the riskiest stock.

f) An **equal-weight portfolio** (25% in each stock) earns the average of the four returns each month. Compute its monthly returns with `rowMeans()`. Is the portfolio's volatility (`sd`) lower than the average stock volatility? (This is the idea of **diversification**.)

````r
# Your code here
````

````r
# ===================== ANSWER =====================
# a)
returns <- matrix(c( 4.2, -1.5,  2.0,  3.1,
                    -2.3,  3.8,  1.2, -0.7,
                     5.1,  2.4, -0.5,  4.6,
                    -1.0, -3.2,  1.8, -2.5,
                     3.3,  4.9,  0.9,  1.4,
                    -0.8,  1.1,  2.2,  2.9),
                  nrow = 6, byrow = TRUE)
rownames(returns) <- c("Jan", "Feb", "Mar", "Apr", "May", "Jun")
colnames(returns) <- c("THYAO", "ASELS", "BIMAS", "GARAN")
returns

# b)
colMeans(returns)
mean(returns[, "THYAO"])

# c)
pos <- returns > 0
pos
colSums(pos)

# d)
max(returns)
which(returns == max(returns), arr.ind = TRUE)   # Mar, THYAO (5.1)
min(returns)
which(returns == min(returns), arr.ind = TRUE)   # Apr, ASELS (-3.2)

# e)
S   <- cov(returns)
S
vol <- sqrt(diag(S))
round(vol, 2)
sd(returns[, "THYAO"])                           # same as vol["THYAO"]
names(which.max(vol))                            # riskiest stock

# f)
port <- rowMeans(returns)
port
sd(port)
mean(vol)
sd(port) < mean(vol)                             # TRUE -> diversification lowers risk
````

# Q11 (Challenge) — Brand Switching and Market Shares with `%*%`

Every month some customers switch between three coffee chains. The **transition matrix** `P` gives the probability that a customer of the *row* brand buys from the *column* brand next month:

| From \ To | Alpha | Beta | Gamma |
|---|---|---|---|
| Alpha | 0.80 | 0.15 | 0.05 |
| Beta | 0.10 | 0.75 | 0.15 |
| Gamma | 0.05 | 0.10 | 0.85 |

Current market shares: Alpha 50%, Beta 30%, Gamma 20%.

**What you need to know**

- Each row of `P` must add up to 1 (every customer goes *somewhere*).
- The **diagonal** of `P` is each brand's **retention rate** (customers who stay).
- Next month's share of brand *j* = Σ over all brands *i* of (share of *i*) × P[i, j]. That is exactly what **matrix multiplication** `%*%` does: `share1 <- share0 %*% P`, where `share0` is a **1 × 3** matrix.

a) Create `P` with names and check that every row sums to 1.

b) Use `diag()` to get the retention rates. Which brand is the most loyal?

c) Create `share0` as a 1 × 3 matrix (`matrix(c(0.5, 0.3, 0.2), nrow = 1)`) and compute next month's shares with `%*%`.

d) Verify Alpha's new share **by hand** with `sum(share0 * P[, "Alpha"])` — it should match (c).

e) Shares after 3 months: `share0 %*% P %*% P %*% P`. Who is gaining, who is losing?

f) Beta launches a loyalty card and its retention rises to 0.85: overwrite `P["Beta", "Beta"]` and lower `P["Beta", "Gamma"]` to 0.05 so the row still sums to 1. Recompute the 3-month shares.

````r
# Your code here
````

````r
# ===================== ANSWER =====================
# a)
P <- matrix(c(0.80, 0.15, 0.05,
              0.10, 0.75, 0.15,
              0.05, 0.10, 0.85),
            nrow = 3, byrow = TRUE)
rownames(P) <- colnames(P) <- c("Alpha", "Beta", "Gamma")
P
rowSums(P)                       # all 1

# b)
diag(P)
names(which.max(diag(P)))        # "Gamma"

# c)
share0 <- matrix(c(0.5, 0.3, 0.2), nrow = 1)
colnames(share0) <- colnames(P)
share1 <- share0 %*% P
share1                           # 0.44 0.32 0.24

# d)
sum(share0 * P[, "Alpha"])       # 0.44

# e)
share3 <- share0 %*% P %*% P %*% P
round(share3, 3)
round(share3 - share0, 3)        # positive = gaining

# f)
P2 <- P
P2["Beta", "Beta"]  <- 0.85
P2["Beta", "Gamma"] <- 0.05
rowSums(P2)
round(share0 %*% P2 %*% P2 %*% P2, 3)
````
